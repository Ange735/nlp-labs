```python
import tensorflow as tf
from tensorflow import keras
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer

sentences = [
    'i love my dog',
    'I, Love my CAT',
    'you love my dog!'
]

tokenizer = Tokenizer(num_words=100) #num_words = max de  mot du vocabulaire qu'on peux avoir apres tokenisation, dans le vocabulaire on garde les num_word premiers mot(par ordre de frequence),ça intervient plus dans l'apprentissage
tokenizer.fit_on_texts(sentences) #num_words n'a pas d'effet sur la fonction du fit

word_index = tokenizer.word_index # le word_index ne tient pas compte du num_words, le num_words c'est ça intervient dans le vocabulaire qu'on construit
#ça scanne le corpus et retient les mots uniques

print(word_index)

#les index de tokenisation sont fixer par ordre de frequence d'apparution, puis par l'ordre d'apparution pour keras/ pour un autre truc ca peut etre différent
```

```python
print("---2. CONVERSIOON EN SEQUENCES(texts_to_sequences)--")

#Ici, num_words a son plein effet :
#seul les mots ayant un index < num_words seront conservés.

sequences= tokenizer.texts_to_sequences(sentences)
print("Sequence générées(avec num_words= 100) :")
print(sentences, ':', sequences)
print("-"*35)

"""Classement par frequence:
{'love': 1, 'my': 2, 'i': 3, 'dog': 4, 'cat': 5, 'you': 6}
"""

#dans le word_index, l'index 0 est reservé, c'est pourquoi il commence par 1. Le 0 est reservé pour un mot bien particulier (a voir par la suite)
```

```python
print("--3. RECONSTRUCTION DU TEXTE(sequences_to_texts)---")

#convertir les sequences numeriques en texte
text_output = tokenizer.sequences_to_texts(sequences)
print("Texte reconstruit :")
print( text_output)
print("-"*35)
```

```python
#BONUS : test avec un Token OOV (out-of-Vocbulary)---
tokenizer_oov = Tokenizer(num_words=3, oov_token="<unk>")
tokenizer_oov.fit_on_texts(sentences)
print("--- 4. BONUS : TEST OOV ---")
print("Nouvel index (num_words=3, avec <unk>):")
print(tokenizer_oov.word_index)
print("-"*35)
```

```python
#Tokenisation d'une nouvelle phrase qui contient des mots hors vocabulaire
test_sentence = ['i love my dog']
sequences_oov = tokenizer_oov.texts_to_sequences(test_sentence)
print("\n Séquence pour 'i love my dog' :")
print(sequences_oov)
print("-"*35)
```

```python
sentences = [
    'I love my dog',
    'I love my cat',
    'You love my dog',
    'Do you think my dog is amazing?'
]

tokenizer = Tokenizer(num_words=100)
tokenizer.fit_on_texts(sentences)

word_index = tokenizer.word_index
print(word_index)

sequences= tokenizer.texts_to_sequences(sentences)
print("Sequence générées(avec num_words= 100) :")
print(sentences, ':', sequences)
print("-"*35)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
#padded = pad_sequences(sequences, maxlen=5) ici le maxlen c'st pour le nombre max de mot qu'on va garder dans chaque phrase, si une phrase depasse le nombre de mot, on fait la troncature
padded= pad_sequences(sequences) # comme ça il pad par defaut en fonction de la plus grnade phrase, il va completer(padder) les autres par des 0 pour atteindre la longueur de la longue phrase
#aussi par defaut il padd en commenceant au debut, pour le faire commencer a la fin on ajoute un argument (truncating='post')
print("\nWord Index=" , word_index)
print("\nSequences=" , sequences)
print("\nPadded Sequences:")
print(padded)
```
